# Harmonise labels

```python
import scanpy as scimport scanpy as sc
sc.set_figure_params(frameon=False, dpi=80, fontsize=12)

from scatlastb_utils.io import read_anndata
sc.set_figure_params(frameon=False, dpi=80, fontsize=12)

from scatlastb_utils.io import read_anndata
```

```python
adata = read_anndata(
    'data/pipeline/load_data/download/Retina_snRNA.h5ad',
    dask=True,
    backed=True,
    verbose=False,
)
adata
```

```python
adata.obs['ann_finest_level'] = adata.obs['cell_type'].astype(str)
```

## AC labels

```python
ac_mask = adata.obs['AC_cluster'].notna()
```

```python
adata.obs.loc[, 'ann_finest_level'].value_counts()
```

```python
adata.obs['AC_cluster'].value_counts()
```

```python
adata.obs['AC_subclass'].value_counts()
```

```python
adata.obs.loc[ac_mask, 'ann_finest_level'] = adata.obs.loc[ac_mask, ['cell_type', 'AC_cluster']].astype(str).apply(', '.join, axis=1)
```

```python
print(adata.obs[ac_mask].value_counts(['AC_cluster', 'ann_finest_level']).to_string())
```

```python
print(adata.obs[ac_mask].value_counts(['AC_subclass', 'ann_finest_level'], sort=False).to_string())
```
