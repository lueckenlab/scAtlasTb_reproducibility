# Transfer annotations to and from unfiltered object

```python
from scatlastb_utils.io import read_anndata
```

```python
cxg_file = 'data/pipeline/load_data/download/HLCAv1_core.h5ad'
unfiltered_file = 'data/input/HLCA_unfiltered.h5ad'
```

```python
obs = read_anndata(cxg_file, obs='obs').obs
obs.index.name = 'index'
obs_unfiltered = read_anndata(unfiltered_file, obs='obs').obs
obs_unfiltered.index.name = 'index'
```

## Columns missing from CxG

```python
cols_not_in_cxg = [
    col for col in obs_unfiltered.columns
    if col not in obs.columns
]
```

```python
cols_not_in_cxg
```

```python
obs_unfiltered[cols_not_in_cxg]
```

```python
obs_unfiltered[cols_not_in_cxg].to_csv('data/input/HLCA_unfiltered_anno.tsv', sep='\t')
```

```python
for col in cols_not_in_cxg:
    obs[col] = obs_unfiltered[col]
```

```python
obs[cols_not_in_cxg]
```

## Columns missing from unfiltered

```python
cols_not_in_unf = [
    col for col in obs.columns
    if col not in obs_unfiltered.columns
]

cols_not_in_unf
```
